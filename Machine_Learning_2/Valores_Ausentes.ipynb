```python
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
# Carga de datos
X_train = [[1,     2],
          [np.nan, 3],
          [7,      8]]
```

```python
# Limpieza de datos: imputacion valores ausentes (modelo)
imputer = SimpleImputer(strategy='mean') # Utilizada la estrategia de la media

#Aplicar los metodos "fit" y "transform" para imputar los valores perdidos de X_train
X_train_clean = imputer.fit_transform(X_train)
print(X_train_clean)
```

```python
# Imputacion de valores ausentes sobre el conjunto de test
X_test = [[np.nan, 10],
         [6,       np.nan],
         [8,       2]]

#imputar los datos del conjunto de test 
X_test_clean=imputer.transform(X_test)
print(X_test_clean)
```
